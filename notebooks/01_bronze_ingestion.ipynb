## Ingestão Bronze 🥉 

```python
from pyspark.sql.functions import *

# Caminho do seu arquivo
file_path = "/Volumes/walmart_sales/default/walmart_sales/Walmart_Sales.csv"

# Leitura do CSV
df = spark.read.csv(
    file_path,
    header=True,
    inferSchema=True
)

# Visualizar dados
display(df)
```

## 💾 Salvando como tabela Delta (Bronze)

```python
df.write.format("delta") \
  .mode("overwrite") \
  .saveAsTable("bronze_walmart_sales")
```

### ✅ Validar

```python
spark.sql("SELECT * FROM bronze_walmart_sales LIMIT 10").show()
```

## ⚙️ Simulando ingestão incremental

```python
from pyspark.sql.functions import col

# Simular novos dados (datas mais recentes)
df_new = df.filter(col("Date") > "2012-01-01")

df_new.write.format("delta") \
  .mode("append") \
  .saveAsTable("bronze_walmart_sales")
```
