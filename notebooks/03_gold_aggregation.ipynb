```python
df_silver = spark.table("silver_walmart_sales")

display(df_silver)
```

```python
from pyspark.sql.functions import year, sum, avg

df_gold = (
    df_silver
    .withColumn("Year", year("Date"))
    .groupBy("Store", "Year")
    .agg(
        sum("Weekly_Sales").alias("Total_Sales"),
        avg("Weekly_Sales").alias("Avg_Sales")
    )
)
```

```python
df_gold.write.format("delta") \
  .mode("overwrite") \
  .saveAsTable("gold_walmart_sales")
```

```python
display(df_gold)
```

```python
df_gold_2 = spark.sql("""
SELECT
  Year,
  SUM(Total_Sales) AS Total_Revenue
FROM gold_walmart_sales
GROUP BY Year
ORDER BY Year
""")

display(df_gold_2)
```
